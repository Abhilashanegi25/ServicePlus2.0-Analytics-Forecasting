```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries loaded successfully")
```

```python
df = pd.read_csv("/Users/abhilashanegi/Downloads/weekly_applications.csv")

print(df.head())
print("\nRows:", len(df))
```

```python
df["week_start"] = pd.to_datetime(df["week_start"])

df = df.sort_values("week_start")

print(df.dtypes)
```

```python
plt.figure(figsize=(12,5))
plt.plot(df["week_start"], df["applications"])
plt.title("Weekly Applications Trend")
plt.xlabel("Week")
plt.ylabel("Applications")
plt.grid(True)
plt.show()
```

```python
import statsmodels

print(statsmodels.__version__)
```

```python
from statsmodels.tsa.arima.model import ARIMA

model = ARIMA(df["applications"], order=(1,1,1))

model_fit = model.fit()

print(model_fit.summary())
```

```python
forecast = model_fit.forecast(steps=4)

print(forecast)
```

```python
future_dates = pd.date_range(
    start=df["week_start"].max() + pd.Timedelta(days=7),
    periods=4,
    freq="W-MON"
)

forecast_df = pd.DataFrame({
    "week_start": future_dates,
    "forecast_applications": forecast.values
})

forecast_df
```

```python
plt.figure(figsize=(12,5))

plt.plot(
    df["week_start"],
    df["applications"],
    label="Historical"
)

plt.plot(
    forecast_df["week_start"],
    forecast_df["forecast_applications"],
    label="Forecast",
    linestyle="--"
)

plt.legend()
plt.title("Weekly Applications Forecast")
plt.xlabel("Week")
plt.ylabel("Applications")

plt.show()
```

```python
forecast_df
```

```python
forecast_df.to_csv(
    "/Users/abhilashanegi/Downloads/forecast_results.csv",
    index=False
)

print("Forecast saved")
```

```python
import pandas as pd

df = pd.read_csv("/Users/abhilashanegi/Downloads/monthly_applications.csv")

df["month_start"] = pd.to_datetime(df["month_start"])

print(df.shape)
df.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.plot(df["month_start"], df["applications"], marker="o")
plt.title("Monthly Applications")
plt.xlabel("Month")
plt.ylabel("Applications")
plt.grid(True)

plt.show()
```

```python
import numpy as np

df["log_apps"] = np.log1p(df["applications"])

plt.figure(figsize=(10,5))
plt.plot(df["month_start"], df["log_apps"], marker="o")
plt.title("Log Transformed Monthly Applications")
plt.xlabel("Month")
plt.ylabel("log(1 + applications)")
plt.grid(True)

plt.show()
```

```python
from statsmodels.tsa.arima.model import ARIMA

model = ARIMA(df["log_apps"], order=(1,1,1))
model_fit = model.fit()

print(model_fit.summary())
```

```python
forecast = model_fit.forecast(steps=6)

forecast_actual = np.expm1(forecast)

print(forecast_actual)
```

```python
future_dates = pd.date_range(
    start=df["month_start"].max() + pd.offsets.MonthBegin(1),
    periods=6,
    freq="MS"
)

forecast_df = pd.DataFrame({
    "month_start": future_dates,
    "forecast_applications": forecast_actual.values
})

forecast_df
```

```python
forecast_df.to_csv(
    "/Users/abhilashanegi/Downloads/monthly_forecast_results.csv",
    index=False
)
```
